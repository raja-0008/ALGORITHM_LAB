# Shortest Path Using Dijkstra Algorithm in Python
Problem Description:
Given a weighted graph with non negative edge weights. Find the shortest distance from a source vertex to all other vertices using Dijkstra algorithm

#Theory
Dijkstra algorithm finds shortest distance from one source node to all other nodes in a weighted graph with non negative weights. Dictionary structure stores nodes as keys and neighbor nodes with weights as values.

##Algorithm Steps

1. Represent graph using dictionary.
2. Set distance of every node as infinity.
3. Set distance of source node as 0.
4. Select node with smallest distance among unvisited nodes.
5. Update distance of neighbor nodes.
6. Repeat until all nodes get visited.

```python
def dijkstra(graph, start):

    distances = {}
    predecessors = {}
    visited = set()

    for node in graph:
        distances[node] = float('inf')
        predecessors[node] = None

    distances[start] = 0

    while len(visited) < len(graph):

        min_node = None
        min_distance = float('inf')

        for node in graph:
            if node not in visited and distances[node] < min_distance:
                min_distance = distances[node]
                min_node = node

        if min_node is None: # This can happen if there are unreachable nodes
            break

        visited.add(min_node)

        for neighbor, weight in graph[min_node].items():
            new_distance = distances[min_node] + weight

            if new_distance < distances[neighbor]:
                distances[neighbor] = new_distance
                predecessors[neighbor] = min_node

    return distances, predecessors
```

```python
def get_shortest_path(predecessors, start, end):
    path = []
    current = end
    while current is not None:
        path.insert(0, current)
        current = predecessors[current]

    if path[0] == start:
        return path
    else:
        return []
```

```python

graph_new = {
    'A': {'B': 1, 'C': 4},
    'B': {'A': 1, 'D': 2, 'E': 5},
    'C': {'A': 4, 'F': 2},
    'D': {'B': 2, 'E': 1},
    'E': {'B': 5, 'D': 1, 'F': 3},
    'F': {'C': 2, 'E': 3}
}

graph_to_use = graph_new

print("Graph being used:")
for node, neighbors in graph_to_use.items():
    print(f"  {node}: {neighbors}")
print("\n")
```

```python


source_nodes = ['A', 'B', 'C', 'D', 'E', 'F']

for start_node in source_nodes:
    print(f"--- Starting Dijkstra from node '{start_node}' ---")
    distances, predecessors = dijkstra(graph_to_use, start_node)

    for target_node in graph_to_use:
        if target_node == start_node:
            print(f"Shortest distance from {start_node} to {target_node}: 0 (Path: [{start_node}])")
            continue

        distance = distances[target_node]
        if distance == float('inf'):
            print(f"Shortest distance from {start_node} to {target_node}: Unreachable")
        else:
            path = get_shortest_path(predecessors, start_node, target_node)
            print(f"Shortest distance from {start_node} to {target_node}: {distance} (Path: {' -> '.join(path)})")
    print("\n")
```

```python
graph = {
'A': {'B': 4, 'C': 2},
'B': {'A': 4, 'C': 5, 'D': 10},
'C': {'A': 2, 'B': 5, 'D': 3},
'D': {'B': 10, 'C': 3}
}

distances, predecessors = dijkstra(graph, 'A') # Unpack the tuple

for node in distances:
    print("Distance from A to", node, "=", distances[node])
```
